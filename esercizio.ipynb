```python
from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self,name,surname,age):
        self.name = name
        self.surname = surname
        self.age = age
    @abstractmethod
    def presentazione(self):
        pass
```

```python
class Studente(Person):
    def __init__(self, name, surname, age, matricola, corso):
        super().__init__(name, surname, age)
        self.matricola = matricola
        self.corso = corso
        self.voti = []

    def aggiungi_voto(self,voto):
        self.voti.append(voto)

    def calcolaMedia(self):
        return sum(self.voti) / len(self.voti)

    def presentazione(self):
        return print(f"Ciao, mi chiamo {self.name} {self.surname}. Ho {self.age} e frequento il corso di {self.corso}")
```

```python
class Dipendente(Person):
    def __init__(self, name, surname, age,azienda,stipendio_mens):
        super().__init__(name, surname, age)
        self.azienda = azienda
        self.stipendio_mens = stipendio_mens
    
    def calcolo_RAL(self):
      return self.stipendio_mens * 12

    def presentazione(self):
            return print(f"Ciao, mi chiamo {self.name} {self.surname}. Ho {self.age}, lavoro in {self.azienda} e guadagno {self.stipendio_mens}")
```

```python
class Docente(Dipendente):
    def __init__(self, name, surname, age, azienda, stipendio_mens,materia,ore_settimanali):
        super().__init__(name, surname, age, azienda, stipendio_mens)
        self.materia = materia
        self.ore_settimanali = ore_settimanali

    def presentazione(self):
         return print(f"Ciao, mi chiamo {self.name} {self.surname}. Ho {self.age}, lavoro in {self.azienda}, insegno {self.materia} e guadagno {self.stipendio_mens} lavorando {self.ore_settimanali} ore a settimana")
```

```python
class Freelance(Dipendente):
    def __init__(self, name, surname, age, azienda, stipendio_mens,pIva,tariffa_oraria,ore_lavorate):
        super().__init__(name, surname, age, azienda = "Autonomo", stipendio_mens = 0)
        self.pIva = pIva
        self.tariffa_oraria = tariffa_oraria
        self.ore_lavorate = ore_lavorate

    def calcolo_compenso(self):
        return self.tariffa_oraria * self.ore_lavorate

    def presentati(self):
        return (f"Sono {self.name} {self.surname}, freelance (P.IVA {self.pIva}), "
                f"tariffa {self.tariffa_oraria} €/h, ")
```
