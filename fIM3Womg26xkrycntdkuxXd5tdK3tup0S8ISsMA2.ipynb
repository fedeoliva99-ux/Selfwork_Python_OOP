## Definizione di una Classe
Una classe rappresenta una porzione generica della realtà, mentre un oggetto è una specifica istanza di quella porzione di realtà.


```python
# Dichiarazione di una classe
class Person:
    # Costruttore per creare un'istanza della classe
    def __init__(self, nome, cognome, eta):
        self.name = nome  # Attributo di istanza
        self.surname = cognome
        self.age = eta

# equivalente in PHP
# class Person {
#  public $name;
#  public $surname;
#     public function construct($nome, $cognome, $eta){
#         $this->name = $nome  # Attributo di istanza
#         $this->surname = $cognome
#         $this->age = $eta   
#     }
# }
#
# $giancarlo = new Person("Giancarlo", "Valente", 50);

# Creazione di istanze della classe Person
giancarlo = Person("Giancarlo", "Valente", 50)
nicola = Person("Nicola", "Milella", 37)
davide = Person("Davide", "Russo", 33)

print(giancarlo)
print(giancarlo.age)

# Accesso agli attributi
# nicola_nome = nicola.name
print( nicola.name)  # Output: Nicola


davide_cognome = davide.surname
print(davide_cognome)  # Output: Russo
```


## Attributi di Classe
Gli attributi di classe sono comuni a tutte le istanze della classe.


```python

class Person:
    # Attributo di classe
    counter = 0

    def __init__(self, nome, cognome, eta):
        self.name = nome  # Attributo d'istanza
        self.surname = cognome
        self.age = eta
        Person.counter += 1

# Creazione di istanze della classe
nicola = Person("Nicola", "Milella", 37)
davide = Person("Davide", "Russo", 33)
g = Person("g", "v", 50)
f = Person("g", "v", 50)


# Accesso agli attributi di classe
print(Person.counter)  # Output: 3
```


## Metodi di Classe e Metodi d'Istanze
Una classe può avere dei metodi d'istanza e dei metodi di classe.


```python
class Person:
    # Attributo di classe
    counter = 0

    def __init__(self, nome, cognome, eta):
        self.name = nome
        self.surname = cognome
        self.age = eta
        Person.counter += 1

    # Metodo d'istanza
    def say_hello(self):
        print(f"Mi chiamo {self.name} {self.surname} e ho {self.age} anni")

    # Metodo di classe (decoratore)
    @classmethod
    def counter_check(cls):
        print(f"Sinora ho creato {cls.counter} persone!")


# Creazione di istanze e uso di metodi
gian = Person('Giancarlo', 'Valente', 50)
nico = Person('Nico', 'Milella', 37)
gian.say_hello()  # Output: Mi chiamo Giancarlo Valente e ho 50 anni
nico.say_hello()
Person.counter_check()  # Output: Sinora ho creato 1 persone!
```



### Abstract Classes
Le classi astratte non possono essere istanziate direttamente ma solo estese.



```python
from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self, nome, cognome, eta):
        self.name = nome
        self.surname = cognome
        self.age = eta
    @abstractmethod
    def say_hello(self):
        pass
        

class Student(Person):
    def __init__(self, nome, cognome, eta, materia):
        super().__init__(nome, cognome, eta)
        self.subject = materia

    def say_hello(self):
        print(f"Ciao a tutti, sono {self.name} {self.surname} e mi piace molto {self.subject}")
        
class Teacher(Person):
    def __init__(self, nome, cognome, eta, materia,stipendio):
        super().__init__(nome,cognome,eta)
        self.subject = materia
        self.salary = stipendio

    def say_hello(self):
        print(f"Mi chiamo {self.name} {self.surname}, ho {self.age} anni, insegno {self.subject} e guadagno {self.salary}")

#p = Person() # ERRORE !!! NON POSSO CREARE UN OGGETTO DA UNA CLASSE ASTRATTA

# Creazione di un'istanza di Teacher
nicola = Teacher('Nicola', 'Milella', 37, 'Python',2000)
mario = Student('Mario', 'Rossi', 37, 'Python')
nicola.say_hello()
mario.say_hello()
```

### `__str__` e `__eq__`
Dunder methods (metodi con doppio underscore) permettono di definire comportamenti speciali per le nostre classi.



```python

class Person:
    def __init__(self, name, surname):
        self.name = name
        self.surname = surname

    def __str__(self):
        return f"Oggetto Person: {self.name} {self.surname}"
        
    # e se fossero piu' oggetti da confrontare come p1 == p2 == p3
    def __eq__(self, other):
        if isinstance(other, Person):
            return self.name == other.name and self.surname == other.surname
        else:
            raise Exception(f"{other} non è un'istanza di Person")

    # def __eq__(self, *others):
    #     print("self:",self)
    #     print("other:",others)

# Creazione e confronto
person1 = Person('Nicola', 'Milella')
person2 = Person('Nicola', 'Milella')
person4 = Person('Nicola', 'Milella')
person3 = Person('Nicola', 'Milella')

print(person1 == person2 == person3 == person4)
```

```python
class Point3D:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def __str__(self):
        return f"x:{self.x} y:{self.y} z:{self.z}"

    # def show(self):
    #     return f"x:{self.x} y:{self.y} z:{self.z}"
        
    def __eq__(self, other):
        if isinstance(other, Point3D):
            return self.x == other.x and self.y == other.y and self.z == other.z
        else:
            raise Exception(f"{other} non è un'istanza di Point3D")


points = [ 
            Point3D(3.44,55,66), 
            Point3D(2,4,6), 
            Point3D(1.77,2,99)
         ]

for p in points:
    print(p)

print(Point3D(3.44,55,66) == Point3D(1.77,2,99))
print(Point3D(3.44,55,66) == Point3D(3.44,55,66))
```

## Attributi/Metodi Privati
In Python non esistono attributi o metodi veramente privati nel senso stretto come in Java, C++ o altri linguaggi fortemente tipizzati.

Python si basa su una filosofia chiamata ***"We are all consenting adults here"***, cioè:

"Non ti impedisco di accedere a qualcosa, ma ti faccio capire che non dovresti."

Gli attributi possono essere resi "privati" aggiungendo doppio underscore `__` davanti al nome.

| Notazione | Significato                               | Accesso esterno | Name mangling |
|-----------|-------------------------------------------|-----------------|---------------|
| `nome`    | Pubblico                                  | ✅              | ❌            |
| `_nome`   | Protetto (convenzione, uso interno)       | ✅              | ❌            |
| `__nome`  | Privato (offuscato tramite name mangling) | ⚠️ (possibile)  | ✅            |


```python

class User:
    def __init__(self, username, email, password):
        self.username = username
        self.email = email
        self.__password = password #_User__password
        self.__phone = "34546655" #_User__phone

    def dimmi_la_tua_password(self):
        self._test()
        print(f"Sei stato molto convincente, La mia password e' {self.__password}")

    def _test(self):
        print("sono privato")
    
    def __test_segreto(self): # _User__test_segreto
        print("sono super privato")
        
    def __str__(self):
        return f"{self.username} {self.email} - password: {self.__password}"

# Creazione di un'istanza di User
nicola = User('Nicola', 'nicola@gmail.com', '123456')
print(nicola)  # Output: Marco marco@gmail.com - password: 123456

# Tentativo di accesso diretto a __password - Fallisce
#try:
print(nicola.__password)
#except AttributeError as e:
   # print(e)  # Output: 'User' object has no attribute '__password'

# di base dovrei chiedere per attributi privati
# nicola.dimmi_la_tua_password()
# # nicola._test()
# nicola._test()
```

Il ***name mangling*** in Python è un meccanismo automatico usato per ***"offuscare"*** i nomi di attributi e metodi che iniziano con due underscore (__) ma non terminano con due underscore.

```python
# Accesso tramite name mangling --- e' una "sporcatura" di Python
# In modo automatico __password viene rinominato in _User__password e questo quindi lo rende accessibile dall'esterno
print(nicola._User__password)  # Output: 123456
# print(nicola._User__phone)
nicola.__test_segreto()
```
